```python
def calcul_taux_mortalite(cas, deces):
    return (deces / cas) * 100
```

```python
taux_algerie = calcul_taux_mortalite(cas1, deces1)
print("Taux mortalité Algérie :", taux_algerie)
```

```python
def niveau_gravite(taux):

    if taux > 1:
        return "Critique"

    elif taux >= 0.5 and taux <= 1:
        return "Modéré"

    else:
        return "Faible"
```

```python
taux = calcul_taux_mortalite(cas1, deces1)
print(niveau_gravite(taux))
```

```python
pays = ["Algérie","France","Italie","USA","Chine"]
cas = [160000,5600000,4200000,33000000,102000]
deces = [4300,107000,127000,590000,4800]

for i in range(len(pays)):

    taux = calcul_taux_mortalite(cas[i], deces[i])
    gravite = niveau_gravite(taux)

    print("Pays :", pays[i])
    print("Cas confirmés :", cas[i])
    print("Décès :", deces[i])
    print("Taux mortalité :", round(taux,2),"%")
    print("Niveau :", gravite)
    print("----------------------")
```

```python
cumul = 0
i = 0

while cumul < 10000000:

    cumul += cas[i]
    print("Pays ajouté :", pays[i])
    print("Cumul des cas :", cumul)

    i += 1
```
